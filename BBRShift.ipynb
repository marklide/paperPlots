```python
#E2 alpha 5.89(30)
#eta E2 0.13(3)
def meanField(T):
    return 8*np.pi**5*sp.constants.k**4*T**4/(15*sp.constants.c**3*sp.constants.h**3*sp.constants.epsilon_0)
def nuBBRE2(T):
    return (-1/(2*sp.constants.h))*(5.89e-40)*(1+0.13)*meanField(T)
```

```python
c = (-1/(2*sp.constants.h))*8*np.pi**5*sp.constants.k**4/(15*sp.constants.c**3*sp.constants.h**3*sp.constants.epsilon_0)
etaE2 = 0.13
deletaE2 = 0.03
etaE3 = 0.0025
deletaE3 = 0.0002
alphaE2 = 5.89e-40
delalphaE2 = 0.3e-40
alphaE3 = 0.888e-40
delalphaE3 = 0.016e-40
def delnuBBR(eta, alpha, T, deleta, delalpha, delT):
    return np.sqrt((delalpha)**2*(c*(1+eta)**2*T**4)**2 + (deleta)**2*(c*alpha*T**4)**2 + (delT)**2*(3*c*alpha*(1+eta)*T**3)**2)

def nuBBR(eta, alpha, T):
    return c*alpha*(1+eta)*T**4
```

```python
#T = 298
#T = 297.65 #our lab
#T = 298.4 #Yb1
T = 299
delT = 1.2
print("E2 shift: ", 1e18*nuBBR(etaE2, alphaE2, T)/fe2)
print("E2 uncertainty: ", 1e18*delnuBBR(etaE2, alphaE2, T, deletaE2, delalphaE2, delT)/fe2)
print()
print("E3 shift: ", 1e18*nuBBR(etaE3, alphaE3, T)/fe3)
print("E3 uncertainty: ", 1e18*delnuBBR(etaE3, alphaE3, T, deletaE3, delalphaE3, delT)/fe3)
```
