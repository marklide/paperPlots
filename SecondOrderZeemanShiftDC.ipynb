```python
from qutip import *
import numpy as np
import scipy as sp
import math
import matplotlib.pyplot as plt
from scipy.special import genlaguerre as L
from scipy.optimize import root_scalar as root
from matplotlib import colormaps as cm
import matplotlib.ticker as ticker
```

```python
fe2 = 688358979309308.22
fe3 = 642121496772645.10
```

```python
h = sp.constants.h
muB = 9.2740100657e-24

nuZ1 = 31275
delnuZ1 = 100
g = 0.59930
delg = 0.00009
B = 31275*sp.constants.h/(0.59930*muB)
delB = np.sqrt(delnuZ1**2*(h/(g*muB))**2 + delg**2*(delnuZ1*h/(g**2*muB))**2)
```

```python
alphaE2 = 52.14e9
delalphaE2 = 0.03e9
alphaE3 = -2.082e9
delalphaE3 = 0.002e9

def nuZ2(alpha, B):
    return alpha*B**2

def delnuZ2(alpha, B, delalpha, delB):
    return np.sqrt(B**4*delalpha**2 + (2*alpha*B)**2*delB**2)
```

```python
print("E2 shift: ", 1e18*nuZ2(alphaE2, B)/fe2)
print("E2 uncertainty: ", 1e18*delnuZ2(alphaE2, B, delalphaE2, delB)/fe2)
print("E2 shift: ", 1e18*nuZ2(alphaE3, B)/fe3)
print("E2 uncertainty: ", 1e18*delnuZ2(alphaE3, B, delalphaE3, delB)/fe3)
```
